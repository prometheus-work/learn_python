```python
message1 = 'Hello world'
print(message1)

#use \' to include single quote in a string
message2 = 'bobby\'s world'
print(message2)

#alternative - use double quotes to include single quote
message3 = "bobby's world"
print(message3)

#use """ triple double quotes to include strings of multiple lines
message4 = """bobby's world was a good
cartoon in the 1900s"""
print(message4)

#print string length use len()
print(len(message1))

##indexing: use square brackets, indexs start from 0
print(message1[0]) #prints out the 0th index which is h

#index error if you try to access an index which doesnt exist

##splicing: pulling single words or letters from string
print(message1[0:5]) #prints starting from 0th index upto but not including the 5th index
                    #first index is the starting point separated by colon is the ending index

print(message1[:5]) #prints ending but not including the 5th index

print(message1[6:]) #prints starting from the 6th index upto the end

##string methods

#.lower() turns the entire string to lowercase
print(message1.lower())

#.upper() turns the entire string to uppercase
print(message1.upper())

#.count() takes an argument and returns how many times that argument appears in the string
print(message1.count("l"))

#.find() takes an argument and returns the index where the argument appears (returns -1 if the argument doesnt exist)
print(message1.find('world'))

#.replace() takes two arguments and replaces the first argument by the second argument
new_message = message1.replace('world', 'universe')
print(new_message)

##concatenate
greeting = 'hello'
name = 'pranav'

message = greeting + name #returns hellopranav
print(message)

message = greeting + ', ' + name #returns hello, pranav
print(message)

#formatted string
message = '{}, {}. Welcome!'.format(greeting, name)
print(message)

message = f'{greeting}, {name}. Welcome' #f string: you can write code within the placeholder like name.upper()
print(message)
```

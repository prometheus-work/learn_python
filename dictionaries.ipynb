```python
##dictionary
# dict = {'key': 'value'}

student = {'name': 'john', 'age': 25, 'courses':['math', 'cs']}
print(student)
#key can be strings and integers (immutable data types)

#print specific value
print(student['name'])

#keyerror when trying to access a key that doesnt exist

#.get("key", "default value") method
print(student.get('phone', 'Not Found')) #returns none/default value if the key doesnt exist

#add values to dictionary
student['phone'] = '555-5555'
print(student.get('phone'))

#edit values
student['name'] = 'jane'
print(student)

#.update() update multiple values at once
student.update({'name': 'john', 'age': '23'})
print(student)

#del delete key value
del student['age']
print(student)

#.pop() removes and returns value
name = student.pop('name')
print(student)
print(name)

#len() no of keys
print(len(student))

#.keys() print all keys
print(student.keys())

#.values() print all values
print(student.values())

#.items() print keys and values
print(student.items())
```

```python
#looping

student = {'name': 'john', 'age': 25, 'courses':['math', 'cs']}

for key, values in student.items():
    print(key, values)
```

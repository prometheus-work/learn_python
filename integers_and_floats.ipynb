```python
#arithmetic operators:
#addition: 3 + 2
#subtraction: 3 - 2
#multiplication: 3 * 2
#division: 3 / 2
#floor division: 3 // 2
#exponent: 3 ** 2
#modulus: 3 % 2

#comparison operators:
#equal: 3==2
#not equal: 3 != 2
#greater than: 3 > 2
#less than: 3 < 2
#greater or equal: 3 >= 2
#less or equal: 3 <= 2
#object identity: is

print(3 // 2) #drops the decimal
print(3 ** 2)
print(3 % 2) #returns remainder after divison

#incrementing values
num =1
num += 1
print(num)
#this can be done with other operators as well
num *=10
print(num)

#abs() gives absolute value
print(abs(-3))

#round(,) rounds value to the nearest int value, you can pass a second argument that tells it how many digits that we want to round to
print(round(3.75, 1))

#type casting
num1 = '100'
num2 = '200'

num1 = int(num1)
num2= int(num2)

print(num1+num2)
```

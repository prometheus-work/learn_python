```python
##list: can modify (mutable)
courses = ['history', 'math', 'phy', 'cs']

print(courses)

#len() no. of values in the list
print(len(courses))

#[index of value] access values in list
print(courses[-1])

#print range of values
print(courses[0:2])

##list methods
#.append() adds value to the end of the list
courses.append('art')
print(courses)

#.insert(index, value)
courses.insert(0, 'bio')
print(courses)

#.extend() join 2 lists
courses_2 = ['fin', 'business']

courses.extend(courses_2)
print(courses)

#.remove() removes value
courses.remove('business')
print(courses)

#.pop() removes last value of the list, also returns the value it removed
popped = courses.pop()
print(courses)
print(popped)

#.reverse() reverses list
courses.reverse()
print(courses)

#.sort() sorts in alphabetical order, numbers in ascending order
courses.sort()
print(courses)

nums = [5, 3, 1, 2]
nums.sort()
print(nums)
#pass reverse=True for the reverse of sort
nums.sort(reverse=True) #gives descending order
print(nums)

#sorted(list) function returns sorted version of list, doesnt alter list
sorted_courses = sorted(courses)
print(sorted_courses)

#min() returns minimum
print(min(nums))

#max() returns maximum
print(max(nums))

#sum() sum of all numbers
print(sum(nums))

#.index() returns index of value inputted
print(courses.index('cs')) #valueerror if the value doesnt exist

#in checks if value is in the list, returns true or false
print('art' in courses)
```

```python
##looping values
for course in courses:
 print(course) #prints all values in courses

#enumerate() returns index and value
for index, course in enumerate(courses, start=1): #start skips the pre values
  print(index, course)

#.join() convert lists into strings
courses_str= ', '.join(courses) #', ' separates each value of courses
print(courses_str)

#.split() turn a string into list
new_list = courses_str.split(', ')
print(new_list)
```

```python
##tuples : cant modify (immutable)

#mutable
list1 = ['history', 'math', 'phy', 'cs']
list2 = list1

print(list1)
print(list2)

list1[0] = 'art'
print(list1)
print(list2)

#immutable
tuple1 = ('history', 'math', 'phy', 'cs')
tuple2 = tuple1

print(tuple1)
print(tuple2)

tuple1[0] = 'art' #gives error because tuple is immutable
print(tuple1)
print(tuple2)

#modify: list
#loop thru, access: tuple
```

```python
##sets: unordered values with no duplicates

cs_courses = {'history', 'math', 'phy', 'cs', 'math'}
print(cs_courses) #only one math is printed - no duplicates

#membership test
print('math' in cs_courses)

##check what values sets share or dont share
art_courses = {'history', 'math', 'art', 'design'}

#.intersection() check which values are common
print(cs_courses.intersection(art_courses))

#.difference() check which values are not in common
print(cs_courses.difference(art_courses)) #what courses are in the cs_courses but not in art_courses

#.union() combine sets
print(cs_courses.union(art_courses)) #what courses are in the cs_courses but not in art_courses
```
